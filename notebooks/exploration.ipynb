```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/telco_churn.csv')
print(df.shape)
print(df.isnull().sum())
```

```python
df['Churn'].value_counts()
```

```python
# TotalCharges is object because of blank entries for new customers
df.dtypes.value_counts()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, col in zip(axes, ['tenure', 'MonthlyCharges', 'TotalCharges']):
    df.groupby('Churn')[col].plot(kind='hist', ax=ax, alpha=0.6, bins=30, legend=True)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
df.groupby('Contract')['Churn'].value_counts(normalize=True).unstack().plot(
    kind='bar', ax=ax
)
ax.set_title('Churn rate by contract type')
ax.set_ylabel('proportion')
ax.tick_params(axis='x', rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Correlation among numeric columns
numeric_cols = df[['tenure', 'MonthlyCharges', 'TotalCharges']]
fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(numeric_cols.corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=ax)
plt.tight_layout()
plt.show()
```

```python
df['Churn_bin'] = (df['Churn'] == 'Yes').astype(int)
churn_by_contract = df.groupby('Contract')['Churn_bin'].mean().round(3)
print(churn_by_contract)
```

```python
avg_tenure = df.groupby('Churn')['tenure'].mean()
print(f'avg tenure (churned):     {avg_tenure["Yes"]:.1f}')
print(f'avg tenure (not churned): {avg_tenure["No"]:.1f}')
```
